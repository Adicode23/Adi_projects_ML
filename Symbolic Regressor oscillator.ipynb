```python
import numpy as np
import pandas as pd
from pysr import PySRRegressor

# 1. Load saved simulation dataset (cols: t, x, v, a)
df = pd.read_csv("oscillator_data.csv") #data of the test case is stored in the csv file
x = df.iloc[:, 1].values
v = df.iloc[:, 2].values
a = df.iloc[:, 3].values

# Stack position (x) and velocity (v) as feature inputs X; acceleration (a) as target y
X = df[["x", "v"]]
y = df["a"]

# 2. Configure the Symbolic Regressor
# We restrict operators to basic arithmetic (+, -, *) to prevent overfitting to unnecessary functions (sin, exp, log)
model = PySRRegressor(
    niterations=50,
    populations=20,
    binary_operators=["+", "-", "*"],
    unary_operators=[],
    elementwise_loss="loss(prediction, target) = (prediction - target)^2",
    model_selection="best"  # Picks the elbow point on the Accuracy vs. Complexity Pareto front
)

# 3. Search for the governing equation
print("Running Symbolic Regression search...")
model.fit(X, y)

# 4. Extract and print the best discovered equation
print("\n--- Rediscovered Governing Equation ---")
print(f"a = {model.sympy()}") #coefficient of v will be (-c/m) and coefficient of x will be (-k/m)
```
